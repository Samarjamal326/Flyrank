## 1. My lane (or freestyle) and why

I am choosing **CTR / Engagement Opportunity Scoring**. I want to identify pages that appear to have meaningful opportunities for better search engagement, so that a content or SEO team can decide which pages deserve further investigation first. This lane is useful because it connects measurable search outcomes such as impressions, clicks, and CTR to a practical prioritization decision rather than simply training a model for its own sake.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print(f"Dataset shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

## 2. The question: decision, action, cost of a wrong call

**Decision:** Which pages appear to have the strongest opportunity for improved search CTR or engagement?

**Who acts:** A content or SEO team can use the ranking to decide which pages should be investigated first.

**Action:** Review the highest-ranked pages and decide whether they need content improvement, title/snippet investigation, better targeting, or simply monitoring.

**Cost of a wrong call:** A false positive could waste time and resources on a page with little improvement opportunity. A false negative could cause a potentially valuable page to be overlooked.

The goal is therefore to provide a **prioritized decision-support ranking**, not an automatic instruction to change every page.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

eligible = df[df["impressions_90d"] >= 100].copy()

zero_click = eligible["clicks_90d"].eq(0)

print(f"Pages with 100+ impressions: {len(eligible):,}")
print(f"Pages with 100+ impressions but 0 clicks: {zero_click.sum():,}")
print(f"Zero-click rate within this group: {zero_click.mean() * 100:.2f}%")

print(df.columns.tolist())
```

## 3. Quick look at the data (2-3 real numbers)

The starter data already contains measurable search-performance outcomes such as impressions, clicks, and CTR. The calculations above provide an initial indication that there are pages receiving search visibility but showing very different engagement outcomes.

I will use these observed patterns to investigate whether a scoring approach can help prioritize pages for further review.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

print(f"Mean CTR: {df['ctr'].mean():.4f}")
print(f"Median CTR: {df['ctr'].median():.4f}")
print(f"Pages with 100+ impressions and 0 clicks: {zero_click.sum():,}")
```

## 4. Careful words: what I can and can't claim

This project can identify **observed patterns and directional opportunities** in the available search-performance data. The eventual score should be treated as **decision-support** for prioritizing pages for investigation.
